**8 · Tu meta de ahorro**



```python
#Sección Resuelve

ahorro = 920
meta = 10000
por_mes = 210
meses = 0

while ahorro<meta:
    ahorro= ahorro+por_mes
    meses+=1
    print(f"Ahorro del mes {meses} es de {ahorro}")
```

```python
print(f"Tardaste {meses} meses en completar tu ahorro")
```
