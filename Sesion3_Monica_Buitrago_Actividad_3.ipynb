**Problema: ¿te alcanza para el café?**

```python
#Sección Resuelve

presupuesto = 100000
precio_cafe =  8500
cafes_semana = 4

gasto_mes = (cafes_semana*precio_cafe)*4
sobra= (presupuesto -gasto_mes)

if sobra >=0:
    alcanza = True
else:
    alcanza =False

print(gasto_mes, sobra, alcanza)
```
